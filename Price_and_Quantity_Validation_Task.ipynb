<b> PRICE AND QUANTITY VALIDATION TASK </b>

```python
# ----- PRICE and QUANTITY VALIDATION -----

product_price = 250
product_quantity = 5
customer_has_coupon = True
customer_budget = 1500
 
subtotal = product_price * product_quantity

discount_eligible = subtotal >= 1000 and customer_has_coupon
 
if discount_eligible:
    discount = 100
else:
    discount = 0
 
final_amount = subtotal - discount
 
within_budget = final_amount <= customer_budget
 
print("Subtotal:", subtotal)
print("Discount Eligible:", discount_eligible)
print("Final Amount:", final_amount)
print("Within Budget:", within_budget)
```
